Example 1: Student marks by department and semester

```python
import pandas as pd

data = {
    "Department": ["CSE", "CSE", "CSE", "DS", "DS", "DS"],
    "Semester": [1, 2, 3, 1, 2, 3],
    "Average_Marks": [76, 81, 79, 82, 85, 88]
}

df = pd.DataFrame(data)

multi = df.set_index(["Department", "Semester"])

print(multi)
```

Selecting data from a MultiIndex

```python
# Select all semester records for CSE
print(multi.loc["CSE"])
```

```python
# Select the average marks for CSE, Semester 2
print(multi.loc[("CSE", 2),"Average_Marks"])
```

```python
# Convert index levels back into ordinary columns
print(multi.reset_index())
```

```python
# Sort index
print(multi.sort_index())

# Swap index levels
print(multi.swaplevel(0, 1))

# Convert inner index level into columns
print(multi["Average_Marks"].unstack())
```

Combining and Merging Data Sets

```python
students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Asha", "Ravi", "Neha", "Imran"],
    "Department": ["CSE", "DS", "CSE", "DS"]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 104, 105],
    "Data_Science_Marks": [86, 91, 78, 88]
})

print("Students Data:")
print(students)

print("Marks Data:")
print(marks)
```

```python
# Keep only IDs present in both datasets
data1 = pd.merge(students, marks, on="Student_ID", how="inner")

print(data1)
```

```python
# Keep every student; unmatched marks become NaN
left = pd.merge(students, marks, on="Student_ID", how="left")

print(left)
```

```python
# Keep all IDs from both datasets
outer = pd.merge(students, marks, on="Student_ID", how="outer")

print(outer)
```

```python
# Keep every record from the right table
right = pd.merge(students, marks, on="Student_ID", how="right")
print(right)
```

```python
jan = pd.DataFrame({
    "Month": ["Jan", "Jan"],
    "Product": ["Pen", "Notebook"],
    "Units": [120, 80]
})

feb = pd.DataFrame({
    "Month": ["Feb", "Feb"],
    "Product": ["Pen", "Notebook"],
    "Units": [150, 95]
})

print(jan)
print(feb)
```

```python
sales = pd.concat([jan, feb], ignore_index=True)
print(sales)
```

```python
student_info = students.set_index("Student_ID")
student_marks = marks.set_index("Student_ID")
joined = student_info.join(student_marks, how="left")
print(joined)
```

Reshaping and Pivoting

```python
sales_long = pd.DataFrame({
    "Product": ["Pen", "Pen", "Notebook", "Notebook"],
    "Month": ["Jan", "Feb", "Jan", "Feb"],
    "Units_Sold": [120, 150, 80, 95]
})

print(sales_long)
```

```python
sales_wide = sales_long.pivot(
    index="Product",
    columns="Month",
    values="Units_Sold"
)

print(sales_wide)
```

```python
transactions = pd.DataFrame({
    "Product": ["Pen", "Pen", "Pen", "Notebook", "Notebook"],
    "Month": ["Jan", "Jan", "Feb", "Jan", "Feb"],
    "Units": [50, 70, 150, 80, 95]
})

print(transactions)
```

```python
summary = transactions.pivot_table(
    index="Product",
    columns="Month",
    values="Units",
    aggfunc="sum",
    fill_value=0
)

print(summary)
```

melt() : Wide format to Long format

```python
wide = pd.DataFrame({
    "Student": ["Asha", "Ravi", "Neha"],
    "Quiz_1": [8, 7, 9],
    "Quiz_2": [9, 8, 10],
    "Quiz_3": [7, 9, 9]
})

print(wide)
```

```python
long = wide.melt(
    id_vars="Student",
    var_name="Quiz",
    value_name="Score"
)

print(long)
```

stack() and unstack()

```python
marks_by_semester = pd.DataFrame({
    "Semester_1": [76, 82],
    "Semester_2": [81, 85]
}, index=["CSE", "DS"])

print(marks_by_semester)
```

```python
#apply stack()
stacked = marks_by_semester.stack()
print(stacked)
```

```python
# Move the inner index level back into columns
print(stacked.unstack())
```

End-to-End Real-World Example: College Performance Analysis

```python
profiles = pd.DataFrame({
    "Student_ID": [201, 202, 203, 204],
    "Name": ["Kiran", "Meera", "Arjun", "Sara"],
    "Department": ["CSE", "DS", "CSE", "DS"]
})

print(profiles)
```

```python
#Create Attendance DataFrame
attendance = pd.DataFrame({
    "Student_ID": [201, 202, 203, 204],
    "Attendance_Percent": [92, 78, 85, 96]
})
print(attendance)
```

```python
results = pd.DataFrame({
    "Student_ID": [201, 202, 203, 204],
    "Final_Marks": [88, 73, 81, 94]
})

print(results)
```

```python
# Combine datasets using Student_ID

analysis_df = profiles.merge(
    attendance,
    on="Student_ID",
    how="left"
)

analysis_df = analysis_df.merge(
    results,
    on="Student_ID",
    how="left"
)

print(analysis_df)
```

```python
# Calculate average attendance and marks by department

dept_summary = analysis_df.groupby("Department")[
    ["Attendance_Percent", "Final_Marks"]
].mean()

print(dept_summary)
```

```python
wide_marks = analysis_df.pivot_table(
    index="Department",
    values="Final_Marks",
    aggfunc="mean"
)

print(wide_marks)
```
